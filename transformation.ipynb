# **Bài thực hành: Sắp xếp và Biến đổi dữ liệu (Sorting & Data Transformation) trong Pandas**

```python
import pandas as pd
import numpy as np

# Khởi tạo tập dữ liệu đơn hàng
data_donhang = {
    'MaDH': ['DH01', 'DH02', 'DH03', 'DH04', 'DH05', 'DH06', 'DH07'],
    'KhachHang': ['An', 'Bình', 'Cường', 'Đông', 'An', 'Hà', 'Nam'],
    'SanPham': ['Laptop', 'Chuột', 'Laptop', 'Bàn phím', 'Màn hình', 'Laptop', 'Chuột'],
    'SoLuong': [1, 2, 1, 3, 2, 1, 4],
    'DonGia': [1500, 25, 1500, 45, 300, 1600, 20],
    'KhuVuc': ['Hà Nội', 'Đà Nẵng', 'Hồ Chí Minh', 'Hà Nội', 'Đà Nẵng', 'Hà Nội', 'Hồ Chí Minh']
}

df = pd.DataFrame(data_donhang)
display(df)
```

## **I. Sắp xếp dữ liệu (Sorting)**

```python
# 1. Sắp xếp giảm dần theo DonGia
df_sort_dongia = df.sort_values(by='DonGia', ascending=False)
print("Sắp xếp theo DonGia giảm dần:")
display(df_sort_dongia)

# 2. Sắp xếp theo KhuVuc (tăng dần) và SoLuong (giảm dần)
df_sort_multi = df.sort_values(by=['KhuVuc', 'SoLuong'], ascending=[True, False])
print("\nSắp xếp theo KhuVuc (tăng dần) và SoLuong (giảm dần):")
display(df_sort_multi)

# 3. Tạo lại chỉ mục mới sau khi sắp xếp
df_sorted = df_sort_multi.reset_index(drop=True)
print("\nBảng dữ liệu sau khi reset index:")
display(df_sorted)
```

## **II. Biến đổi dữ liệu (Data Transformation)**

```python
# 1. Tạo cột ThanhTien
df_sorted['ThanhTien'] = df_sorted['SoLuong'] * df_sorted['DonGia']

# 2. Phân loại cột MucGia dựa vào DonGia
def phan_loai_gia(dongia):
    if dongia >= 1000:
        return 'Cao'
    elif 300 <= dongia < 1000:
        return 'Trung bình'
    else:
        return 'Thấp'

df_sorted['MucGia'] = df_sorted['DonGia'].apply(phan_loai_gia)

# 3. Chuẩn hóa tên khách hàng thành chữ in hoa
df_sorted['KhachHang'] = df_sorted['KhachHang'].str.upper()

display(df_sorted)
```

## **III. Tổng hợp và phân tích dữ liệu (Groupby)**

```python
# 1. Tổng doanh thu (ThanhTien) theo từng KhuVuc
doanh_thu_khuvuc = df_sorted.groupby('KhuVuc')['ThanhTien'].sum().reset_index()
print("Tổng doanh thu theo từng Khu Vực:")
display(doanh_thu_khuvuc)

# 2. Số lượng đơn hàng của từng SanPham
don_hang_sanpham = df_sorted.groupby('SanPham').size().reset_index(name='SoLuongDonHang')
print("\nSố lượng đơn hàng của từng Sản Phẩm:")
display(don_hang_sanpham)

# 3. Giá trị đơn hàng trung bình theo từng Khách hàng
gia_tri_tb_khach = df_sorted.groupby('KhachHang')['ThanhTien'].mean().reset_index(name='GiaTriTrungBinh')
print("\nGiá trị đơn hàng trung bình của mỗi Khách Hàng:")
display(gia_tri_tb_khach)
```

Doanh thu theo khu vực:
- Hà Nội dẫn đầu với tổng doanh thu 3,235.
- Hồ Chí Minh đứng thứ hai với 1,580.
- Đà Nẵng đứng cuối với 650.

Số lượng đơn hàng theo sản phẩm: Laptop là sản phẩm phổ biến nhất với 3 đơn hàng, tiếp theo là Chuột (2 đơn), Màn hình và Bàn phím (mỗi loại 1 đơn).

Giá trị trung bình mỗi khách hàng: Khách hàng HÀ có giá trị đơn hàng trung bình cao nhất (1,600), theo sau là CƯỜNG (1,500). Khách hàng BÌNH có giá trị trung bình thấp nhất (50).
